#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

> **My answer:** Just like KNN regression/classification, K-Means clustering (often) relies on Euclidean distance to determine which cluster an observation should be sorted into. Euclidean distance is sensitive to different scales as it is the square root of the *sum* of the squared differences between related coordinates of two vectors. In the case of K-Means, the distance between observation $X$ with $k$ features to a centroid $C$ is calcuated using $\sqrt{\sum_{i=1}^{k} (X_i - C_i)^2}$. If say, feature $1$ involves a very large scale, but feature $1$ has a small scale, then $(X_2 - C_2)^2$ is much more likely to be larger than $(X_1 - C_1)^2$. This assigns greater significance to feature $2$ than feature $1$ in determining which cluster $X$ is closest to, when in reality, feature $1$ is more important feature in clustering the data. Scaling the data ensures that a feature's significance in clustering the data reflects reality rather than be due to its scale. 

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# initializing KMC with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=2025)

# fitting to scaled wine data
clusters = kmeans.fit(scaled_wine_df)

# adding labels
clustered_wine_data = wine_df.copy()
clustered_wine_data['Cluster'] = clusters.labels_

# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

> **My answer:** One common method for calculating the optimal number of clusters is known as the *elbow method*. First, we can use the Within-Cluster Sum of Squared Distances (WSSD) as an overall indicator of how close observations within each cluster are to said cluster's centroid given $k$ clusters. If the number of clusters adequately groups the data, then observations within a cluster should be close together (i.e. near the centroid), thus leading to a smaller WSSD compared to a clustering with a less adequate number of clusters. But that is not to say that the larger number of clusters, the better; at a certain point, too many clusters means that we lose meaningful groupings since observations are just being grouped into clusters consisting of 1-2 data points (in the extreme case). The WSSD also stops decreasing by a meaningful amount at cuch a point. So the idea is to run KMC with a reasonable range of $k$ values, calculate WSSD per KMC and plotting the resulting WSSD to identify the "elbow" which is the point at which the WSSD ceases to decrease by meaningful levels as the number of clusters increase. 

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# calculate the mean of colour_intensity
mean_color_intensity = wine_df["color_intensity"].mean()

# set seed
np.random.seed(123)

# initialize list for storing bootstrapped means
bootstrapped_means = []

for i in range(10000):
    sample = wine_df['color_intensity'].sample(frac=1, replace=True) # sample with replacement
    avg = sample.mean() # calculate mean
    bootstrapped_means.append(avg) # add to bootstrapped_means

# convert list to dataframe
bootstrapped_means_df = pd.DataFrame({"means": bootstrapped_means})

lower_bound = bootstrapped_means_df['means'].quantile(0.05)
upper_bound = bootstrapped_means_df['means'].quantile(0.95)

# Display the result
print(f"Mean of Color Intesity: {mean_color_intensity}")
print(f"90% Confidence Intrval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

>  **My answer:** A mean is a point estimate regarding a sample which may vary with the particular sample we take. Furthermore, we are often interested in characteristics of the larger population, but do not have the ability to take multiple samples. Without being able to take new wine samples, bootstrapping helps us understand the range of values that the mean colour intensity can take on if we were to take a new sample of wines. 

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

>  **My answer:** The confidence interval helps capture the majority of the bootstrapped means (while omitting potential outliers), thus providing a relatively stable range of variability that the mean can potentially take on with further resampling. 

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

>  **My answer:** the mean of colour intensity appears to be stable and reliable as it lies relatively centered within the confidence interval and the interval also seems to be relatively narrow. However, it is difficult to assess whether the confidence interval is very narrow given there is nothing to compare it against. The only other interval we can compare it to is the range of the sample (see below). Note that it is theoretically possible for a bootstrapped sample to consist only of the minimum value or maximum value, so the range can be viewed as the smallest and largest values that a bootstrapped mean could potnetially be. When we compare the range to the CI, we can note that the range is approximately $CI \pm 0.3$ and one would need more information regarding colour intensity to determine if this differnce is considered large or small. 

```python
minimum = bootstrapped_means_df['means'].min()
maximum = bootstrapped_means_df['means'].max()
print(f"The range of colour intensity is [{minimum}, {maximum}]")
```
